# Cross-modal wrap-up

Everyone spent the afternoon on a different corner of the same disease. This last part is a **discussion, not a scoreboard.**

> There is deliberately no shared results file and no comparison between your numbers and anybody else's. The numbers are not comparable — different cohorts, different sample sizes, different label definitions, different difficulty — and lining them up on one bar chart would teach exactly the wrong lesson. **That incomparability is itself the point.**

What *is* comparable is what each modality can and cannot tell you.


## 1 · Report back — three sentences each

Whoever did each module, say:

1. **What was your data, and how much of it was real?**
2. **What was the one flaw that would have fooled you?**
3. **What is the single figure you would show somebody?**

| Module | The flaw it was built around |
|---|---|
| **A** MRI | Subject-level leakage — one patient, several scans, split across train and test |
| **C** biomarkers | Label leakage through the MMSE, and missingness that tracks severity |
| **D** confounding | Age is entangled with diagnosis; education's effect reverses when you stratify |
| **E** clinical records | Competing risk of death, reverse causation, immortal time |
| **F** genetics | Multiple testing, population stratification, and scores that do not transfer |
| **G** transcriptomics | p ≫ n, and cell-composition change masquerading as gene regulation |
| **H** chemistry | Scaffold leakage — the exact twin of module A's problem, in another discipline |

**Notice the pairs.** A and H are the same mathematical problem in two fields that never cite each other. C and D are the same dataset asked two different questions. That is the most transferable thing in the whole day: the failure modes are *general*, and recognising one in a field you know nothing about is a genuinely useful skill.


## 2 · Cost, invasiveness and what each modality is actually for

| Modality | Cost | What the patient goes through | Used in practice for |
|---|---|---|---|
| Blood biomarkers | €50–300 | A blood draw | Emerging: triage before specialist referral |
| Genetics | ~€50, once, forever | A cheek swab | Research and trial stratification, not clinical care |
| Clinical risk factors | Free | A conversation | Population prevention advice |
| Structural MRI | €300–800 | 30 minutes in a scanner | Confirming a diagnosis, excluding other causes |
| PET imaging | €2000–4000 | Injection, scan, radiation dose | Confirming amyloid before an expensive therapy |
| Brain transcriptomics | — | Only possible after death | Understanding mechanism |
| Compound screening | — | No patient involved | Deciding what to synthesise |

### Discuss

**a)** You have to screen an entire national population aged 65+. Which modality, and what happens to the people your test gets wrong?

**b)** A patient in front of you has memory complaints. Which modality do you reach for, and in what order?

**c)** Which of today's modules was trying to **predict** something, and which was trying to **understand** something? Does a lower AUROC mean a less valuable piece of work?


## 3 · The four things worth remembering next week

**1. The split is the experiment.** Nearly every module had a way to accidentally show the model something it should not have seen — the same patient, the same molecule, the same cognitive test that defined the label. The number a model reports is a statement about how you split the data, not only about the model.

**2. A baseline is not a formality.** In several modules a single well-chosen number came close to the fanciest model. If a method cannot beat one number, it has not earned its complexity — and 'we used deep learning' is not a result.

**3. Ask what the score is made of.** Module D took a respectable AUROC apart and found age underneath. Shapley values in C, E, F and H did the same thing per patient. Interpretability does not make a model correct; it makes it *arguable*, which is the precondition for finding out whether it is correct.

**4. Every model here optimises the wrong thing, slightly.** AUROC, gene lists, enrichment factors — none of them is "the patient got better". Choosing what to predict is a scientific and ethical decision that happens before any code is written, and the BACE1 story in module H is what it looks like when that decision goes wrong at a cost of billions.
